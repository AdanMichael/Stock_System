```python
%matplotlib inline
import pandas as pd
import datetime
import warnings
import statsmodels.api as sm
import matplotlib.pylab as plt
from matplotlib.pylab import style
from matplotlib.font_manager import FontProperties
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.graphics.tsaplots import plot_acf,plot_pacf
from statsmodels.tsa.stattools import adfuller
import warnings
warnings.filterwarnings('ignore')
```

```python
#股票数据
# index_col：我们在读取文件之后所得到的DataFrame的索引默认是0、1、2……
data = pd.read_csv('../data/Day.csv', index_col=0, parse_dates=[0] ,skiprows=range(1,2170))  #
stock_train= data['c'].dropna()
```

```python
stock_train.plot()
plt.title('stock close')
plt.xlabel('time')
plt.ylabel('close price')
plt.show()
```

```python
#使用adf检验的方法确定股票价格是否走势平稳，不平稳时使用差分法进行处理
dftest=adfuller(stock_train)
#下标为1时表示的是 p的值 adf 检验中我们认为一般p的值小于0.05时可以认为是走势平稳
print('数据未进行差分前的adf检验p值：',dftest[1])
```

```python
#确定模型参数d
stock_diff_1=stock_train.diff(1)
#df.dropna 表示过滤缺失值
stock_diff=stock_diff_1.dropna()
print("一阶差分处理之后的数据分布：")
print(stock_diff)
```

```python
#再次检验adf中p的值
dftest=adfuller(stock_diff)
print('adf检验之后的p的值：',dftest[1])
```

```python
#绘制差分之后的图形
stock_diff.plot()
plt.title('First-order difference') #一阶差分
plt.xlabel('time')
plt.show()
```

```python
#绘制自相关系数的图形 找出模型参数中的p值
plt.figure(dpi=600)
acf = plot_acf(stock_diff, lags=20)
plt.title("ACF")
plt.xlabel("Order") #阶数
plt.ylabel("Autocorrelation coefficient")#自相关系数
plt.show()
```

```python
#绘制偏自相关系数的图形 找出模型参数里的q值
pacf = plot_pacf(stock_diff, lags=20)
plt.title("PACF")
plt.xlabel("Order")
plt.ylabel("Partial autocorrelation coefficient")# 偏自相关系数
plt.show()#展示
```

```python
#使用aic 准则法   确定p和q的值
res=sm.tsa.arma_order_select_ic(stock_diff,max_ar=5,max_ma=5,ic=['aic'])
print(res.aic_min_order)
```

```python
model = ARIMA(stock_train, order=(2, 1, 4))      #ARIMA算法模型
re = model.fit()
pred = re.predict(5720, 5741, dynamic=True,type='levels')      #预测
print(pred)
```

```python
s_true=stock_train[5720:5741]
print(s_true)
```

```python
error_1=pow(s_true-pred,2)
MSE1=error_1.sum()/len(pred)
print('模型均方根误差为：',MSE1)
```

```python
plt.figure()
plt.xticks(rotation=45)
plt.plot(pred)
plt.plot(stock_train[5413:])
plt.show()
```
